#### Import Packages

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

#### Import Dataset

```python
df = pd.read_csv("/Users/ann/Documents/ddfg/resident-list-district-6.csv", low_memory=False)
```

#### Filter age 60 and above

```python
df['DOB'] = pd.to_datetime(df['DOB'], format='%m/%d/%Y')

today = datetime.today()
df['Age'] = df['DOB'].apply(lambda dob: today.year - dob.year - ((today.month, today.day) < (dob.month, dob.day)))

df['Age_60_or_Above'] = df['Age'] >= 60

df
```

#### Correct zip code format

```python
df['Zip'] = df['Zip'].astype(str).str.zfill(5)

df
```

#### Filter D6 by zip codes

```python
df = df[df['Zip'].isin(["02130", "02131", "02132"])]

df
```

#### Drop unnecessary columns

```python
df.drop(columns=['Last Name', 'First Name', 'MI', 'Title', 'DOB', 'Occupation', 'Reg Date', 'Res ID'], inplace=True)

df
```

#### Save to CSV

```python
df.to_csv("/Users/ann/Documents/ddfg/resident-list-district-6-clean.csv", index=True)
```
